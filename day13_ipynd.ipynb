1. لماذا نرسم؟

في مجموعات Anscombe’s quartet الأربع تطلع الأرقام الملخِّصة (المتوسط والانحراف والارتباط) تقريباً نفسها، لكن أول ما ترسمها تشوف أربع قصص مختلفة: علاقة خطية عادية، ومنحنى، ونقطة شاذة وحدة تسحب الخط كله، ومجموعة أغلب نقاطها على قيمة وحدة.

مثال من حياتي: لو قيل لي إن متوسط درجات الفصل 75، ما أعرف إذا كان أغلب الطلاب قريبين من 75، أو نصفهم جابوا 100 والنصف الثاني 50. الرقم نفسه والصورة مختلفة تماماً.

معناها لمحلل الداتا: الرقم الواحد يختصر الداتا ويخفي أشياء مهمة مثل شكل التوزيع والقيم الشاذة. فالأفضل أرسم الداتا أول، وبعدها أعتمد على الأرقام.

2. اختر الرسمة
السؤال	الرسمة	لماذا؟
كيف تغيّرت مبيعات المقهى خلال 12 شهراً؟	Line	الداتا مرتبة بالزمن، والخط يبيّن لي الصعود والنزول بوضوح
أي فرع باع أكثر هذا الشهر من 5 فروع؟	Bar	أقارن بين فروع منفصلة، وأعرف الأعلى من طول العمود مباشرة
كيف تتوزع أعمار 500 عميل؟	Histogram	عندي متغير رقمي واحد (العمر) وأبي أشوف وين يتجمع أغلب العملاء
هل يزيد الإيجار كلما زادت مساحة الشقة؟	Scatter	أختبر علاقة بين رقمين، وكل نقطة تمثل شقة وحدة
هل في رواتب الموظفين قيم شاذة؟	Box plot	القيم الشاذة تطلع نقاط بعيدة عن الصندوق، فأكتشفها بسرعة
3. أجزاء الرسمة
Figure: هي اللوحة الكاملة اللي تحوي كل شيء، وتقدر تحوي رسمة وحدة أو أكثر.
Axes: هي مكان الرسم نفسه، فيه المحورين والبيانات. وفي الـ Figure الواحد ممكن يكون أكثر من Axes، مثل ما يصير في subplots.

وظيفة كل جزء:

العنوان: يقول للقارئ إيش تعرض الرسمة.
عنوانا المحورين: يبيّنون إيش يقيس كل محور وبأي وحدة (ريال، شهر، عدد...).
Legend: يوضح كل لون أو خط يمثل أي مجموعة، لما يكون في الرسمة أكثر من مجموعة.

لو أرسلت لمديري رسمة بدون هذي الأجزاء: ما راح يعرف عن إيش الرسمة ولا وش تقيس الأرقام ولا وحدتها، ولا أي خط يمثل إيش. يضطر يسألني أو يخمّن، وممكن يفهمها غلط، وتصير الرسمة ما تفيد.

4. رسمة مضلِّلة

الخدعة: محور Y ما يبدأ من الصفر، يبدأ من حوالي 93. فيبان عمود 96% كأنه ضعف عمود 94%، والفرق الحقيقي نقطتين مئويتين بس.

كيف أصلحها:

أخلي محور Y يبدأ من الصفر عشان ارتفاع الأعمدة يطابق قيمها الحقيقية.
أكتب القيمة (96% و94%) فوق كل عمود.
أحط عنوان صادق، مثل: «رضا العملاء: 96% هذا العام مقابل 94% العام الماضي».

بعد التصحيح يبان العمودان شبه متساويين، وهذا هو حجم التحسن الحقيقي.



## Task 2
### رسومات المقهى

```python
import matplotlib.pyplot as plt
import numpy as np
from collections import Counter

days = ["Sat", "Sun", "Mon", "Tue", "Wed", "Thu", "Fri"]
latte = [42, 38, 35, 40, 44, 58, 61]
espresso = [30, 28, 33, 31, 29, 35, 22]

drinks = ["Latte", "Espresso", "Mocha", "Tea"]
ratings = [5, 4, 4, 5, 3, 5, 4, 2, 5, 4, 4, 5, 3, 4, 5, 5, 1, 4, 4, 5,
           3, 4, 5, 4, 2, 5, 4, 4, 3, 5, 4, 5, 5, 4, 3, 4, 5, 4, 4, 5]

temp = [18, 22, 25, 27, 30, 33, 35, 38, 40, 41]
iced = [5, 9, 12, 15, 20, 26, 30, 35, 39, 38]
```

**شرح الكود:** استوردت `matplotlib.pyplot` للرسم، و`numpy` لتحديد حدود أعمدة الـ Histogram، و`Counter` لعدّ تكرار كل تقييم. بعدها كتبت الداتا كقوائم: أيام الأسبوع ومبيعات الـ Latte والـ Espresso، وأسماء المشروبات، وتقييمات العملاء، والحرارة مع مبيعات القهوة المثلجة.


### 1. Line: مبيعات Latte و Espresso

```python
plt.figure(figsize=(8, 4))
plt.plot(days, latte, marker="o", label="Latte")
plt.plot(days, espresso, marker="s", label="Espresso")
plt.title("Daily Sales: Latte vs Espresso")
plt.xlabel("Day")
plt.ylabel("Cups sold")
plt.legend()
plt.show()
```

**شرح الكود:** `plt.figure(figsize=(8, 4))` تنشئ لوحة بعرض 8 وارتفاع 4. `plt.plot` ترسم خط لكل مشروب، و`marker` يحط علامة على كل نقطة (دائرة أو مربع)، و`label` هو الاسم اللي يظهر في الـ Legend. `plt.title` و`plt.xlabel` و`plt.ylabel` للعنوان وعنواني المحورين، و`plt.legend()` تعرض مفتاح الألوان، و`plt.show()` تعرض الرسمة.

**ماذا تقول هذه الرسمة؟** مبيعات الـ Latte ترتفع بوضوح في نهاية الأسبوع، وأعلى يوم هو الجمعة بـ 61 كوب. أما الـ Espresso فيبقى بين 28 و35 تقريباً من السبت إلى الخميس، ثم ينزل يوم الجمعة إلى 22، وهو أقل رقم له، في نفس اليوم اللي وصل فيه الـ Latte لقمته. ممكن بعض الزبائن تحوّلوا من Espresso إلى Latte، لكن هذا مجرد احتمال ولازم أتأكد منه ببيانات أكثر.

### 2. Bar: إجمالي الأسبوع لكل مشروب

```python
totals = [sum(latte), sum(espresso), 145, 96]
print(dict(zip(drinks, totals)))

plt.figure(figsize=(7, 4))
plt.bar(drinks, totals, color=["#4C72B0", "#55A868", "#C44E52", "#8172B2"])
for i, v in enumerate(totals):
    plt.text(i, v + 3, str(v), ha="center")
plt.title("Weekly Total Sales by Drink")
plt.xlabel("Drink")
plt.ylabel("Cups sold")
plt.show()
```

**شرح الكود:** `sum(latte)` و`sum(espresso)` تحسب إجمالي الأسبوع للمشروبين، وأضفت Mocha = 145 و Tea = 96 مثل ما في المهمة. `print(dict(zip(...)))` تطبع كل مشروب مع إجماليه عشان أتأكد من الأرقام. `plt.bar` ترسم الأعمدة. وحلقة `for` مع `enumerate` تمر على كل عمود، و`plt.text(i, v + 3, str(v), ha="center")` تكتب الرقم فوق العمود: `i` مكانه الأفقي، و`v + 3` ارتفاعه مع رفع بسيط، و`ha="center"` تخليه في المنتصف.

**ماذا تقول هذه الرسمة؟** الـ Latte هو الأعلى مبيعاً في الأسبوع بـ 318 كوب، وبعده الـ Espresso بـ 208، ثم الـ Mocha بـ 145، وأقلهم الـ Tea بـ 96. يعني الـ Latte لحاله يبيع أكثر من ثلاثة أضعاف الـ Tea.

### 3. Histogram: توزيع التقييمات

**توقعي قبل التشغيل:** توقعي اتوقع ان تتتركز التقيمات في الارقام لعاليه4و5 لان  غالبيه تقيمات العملاءعادة تكون ايجابيه

```python
plt.figure(figsize=(7, 4))
plt.hist(ratings, bins=np.arange(0.5, 6.5, 1), edgecolor="black")
plt.title("Distribution of Customer Ratings")
plt.xlabel("Rating (1-5)")
plt.ylabel("Number of customers")
plt.show()

print(sorted(Counter(ratings).items()))
```

**شرح الكود:** `plt.hist` ترسم توزيع التقييمات. `bins=np.arange(0.5, 6.5, 1)` تجعل كل تقييم (1 إلى 5) في عمود لحاله، لأن الحدود 0.5 و1.5 و2.5 وهكذا تخلي الرقم الصحيح في منتصف عموده. `edgecolor="black"` يرسم حدود سوداء بين الأعمدة. `Counter(ratings)` تعدّ كم مرة تكرر كل تقييم، و`sorted` ترتبها من 1 إلى 5.

**ماذا تقول هذه الرسمة؟** أغلب العملاء راضين: التقييم الأكثر تكراراً هو 4 (17 مرة)، وبعده 5 (15 مرة)، يعني 32 عميل من 40 أعطوا 4 أو 5. التقييمات المنخفضة (1 و2) قليلة جداً، 3 عملاء بس. **نتيجة توقعي:** (اكتب هنا هل توقعك طلع صح أو غلط، والجواب الصحيح هو 4).

### 4. Scatter: الحرارة والقهوة المثلجة

```python
plt.figure(figsize=(7, 4))
plt.scatter(temp, iced, color="orange")
plt.title("Temperature vs Iced Coffee Sales")
plt.xlabel("Temperature (°C)")
plt.ylabel("Iced coffee sold")
plt.show()
```

**شرح الكود:** `plt.scatter(temp, iced)` ترسم نقطة لكل يوم: موقعها الأفقي هو الحرارة والعمودي هو مبيعات القهوة المثلجة. `color="orange"` يغيّر لون النقاط، والباقي عنوان وعنواني المحورين.

**ماذا تقول هذه الرسمة؟** كل ما ارتفعت الحرارة زادت مبيعات القهوة المثلجة، يعني علاقة طردية قوية: من 5 أكواب عند 18 درجة إلى 39 كوب عند 40 درجة. آخر نقطة (41 درجة) نزلت شوي إلى 38، فممكن المبيعات توقف عن الزيادة عند الحرارة العالية جداً، لكن نقطة وحدة ما تكفي للجزم.

### 5. Subplots: الداشبورد

```python
fig, axs = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle("Cafe Dashboard", fontsize=16)

# Line
axs[0, 0].plot(days, latte, marker="o", label="Latte")
axs[0, 0].plot(days, espresso, marker="s", label="Espresso")
axs[0, 0].set_title("Daily Sales")
axs[0, 0].set_xlabel("Day")
axs[0, 0].set_ylabel("Cups sold")
axs[0, 0].legend()

# Bar
axs[0, 1].bar(drinks, totals, color=["#4C72B0", "#55A868", "#C44E52", "#8172B2"])
for i, v in enumerate(totals):
    axs[0, 1].text(i, v + 3, str(v), ha="center")
axs[0, 1].set_title("Weekly Total by Drink")
axs[0, 1].set_xlabel("Drink")
axs[0, 1].set_ylabel("Cups sold")

# Histogram
axs[1, 0].hist(ratings, bins=np.arange(0.5, 6.5, 1), edgecolor="black")
axs[1, 0].set_title("Ratings Distribution")
axs[1, 0].set_xlabel("Rating (1-5)")
axs[1, 0].set_ylabel("Customers")

# Scatter
axs[1, 1].scatter(temp, iced, color="orange")
axs[1, 1].set_title("Temperature vs Iced Coffee")
axs[1, 1].set_xlabel("Temperature (°C)")
axs[1, 1].set_ylabel("Iced coffee sold")

plt.tight_layout()
plt.savefig("cafe_dashboard.png", dpi=150)
plt.show()
```

**شرح الكود:** `plt.subplots(2, 2)` ترجع Figure واحد فيه 4 Axes مرتبة في صفين وعمودين، وأوصل لكل واحد بالفهرس مثل `axs[0, 0]` (الصف الأول، العمود الأول). `fig.suptitle` تحط العنوان الرئيسي فوق الكل. مع الـ Axes أستخدم `set_title` و`set_xlabel` و`set_ylabel` بدل `plt.title` وأخواتها. `plt.tight_layout()` تمنع تداخل الرسمات، و`plt.savefig("cafe_dashboard.png", dpi=150)` تحفظ الصورة في نفس مجلد الـ Notebook، و`dpi=150` يعطيها وضوح جيد. حطيت `savefig` قبل `plt.show()` لأن `show` ممكن تفرّغ الرسمة قبل الحفظ.

**ماذا تقول هذه الرسمة؟** الداشبورد يجمع الصورة الكاملة للمقهى: الـ Latte هو الأكثر مبيعاً وقمته يوم الجمعة، والعملاء راضين في الغالب (أغلب التقييمات 4 و5)، والقهوة المثلجة تزيد مع ارتفاع الحرارة. يعني المقهى يقدر يجهّز كمية أكبر من الـ Latte والمثلجات في الأيام الحارة ونهاية الأسبوع.

**توقعي قبل التشغيل:** أتوقع أن التقييم الأكثر تكراراً هو 4، لأن أغلب العملاء راضين لكن قليل منهم يعطي العلامة الكاملة.

### التأمين الطبي بعيني

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("insurance.csv")
df = df.drop_duplicates()

print(df.shape)
df.head()
```

**شرح الكود:** `pd.read_csv` تقرأ الملف، و`drop_duplicates()` تحذف الصف المكرر زي ما طلبت المهمة. `df.shape` تعرض عدد الصفوف والأعمدة للتأكد أن الحذف تم، و`df.head()` تعرض أول 5 صفوف عشان أتأكد من أسماء الأعمدة.

### 1. Histogram للتكاليف مع المتوسط والوسيط

```python
mean_c = df["charges"].mean()
median_c = df["charges"].median()
print(f"Mean: {mean_c:,.0f} | Median: {median_c:,.0f}")

plt.figure(figsize=(8, 4))
plt.hist(df["charges"], bins=30, edgecolor="black", color="#8172B2")
plt.axvline(mean_c, color="red", linestyle="--", label=f"Mean = {mean_c:,.0f}")
plt.axvline(median_c, color="green", linestyle="-", label=f"Median = {median_c:,.0f}")
plt.title("Distribution of Insurance Charges")
plt.xlabel("Charges ($)")
plt.ylabel("Number of people")
plt.legend()
plt.show()
```

**شرح الكود:** `mean()` و`median()` تحسب المتوسط والوسيط. `plt.hist(..., bins=30)` تقسم التكاليف إلى 30 فئة. `plt.axvline` ترسم خط عمودي عند القيمة، وأعطيت كل خط لون ونمط مختلف و`label` عشان يظهر في الـ Legend.

**ماذا تقول هذه الرسمة؟** التوزيع ملتوي لليمين: أغلب الناس تكاليفهم منخفضة (أقل من 15,000 تقريباً)، وفي ذيل طويل لأقلية تكاليفها عالية جداً. المتوسط (حوالي 13,270) أكبر من الوسيط (حوالي 9,400) لأن القيم العالية تسحب المتوسط لفوق، بينما الوسيط ما يتأثر فيها. يعني الوسيط يمثل الشخص "العادي" أفضل من المتوسط هنا.

### 2. Scatter: العمر مقابل التكاليف حسب التدخين

```python
plt.figure(figsize=(8, 5))
for status, color in [("no", "tab:blue"), ("yes", "tab:red")]:
    sub = df[df["smoker"] == status]
    plt.scatter(sub["age"], sub["charges"], color=color, alpha=0.6, label=f"Smoker: {status}")
plt.title("Age vs Charges by Smoking Status")
plt.xlabel("Age (years)")
plt.ylabel("Charges ($)")
plt.legend()
plt.show()
```

**شرح الكود:** حلقة `for` تمر على الحالتين (غير مدخن ومدخن)، وفي كل مرة `df[df["smoker"] == status]` تفلتر الصفوف الخاصة بهذي الحالة، وأرسمها بلون مختلف. `alpha=0.6` يخلي النقاط شبه شفافة عشان أشوف النقاط المتراكبة.

**ماذا تقول هذه الرسمة؟** أشوف 3 «شرائط» تقريباً تصعد مع العمر. الشريط الأدنى فيه أغلب غير المدخنين وتكاليفهم أقل. الشريط الأعلى (فوق 30,000 تقريباً) كله مدخنين. والشريط الأوسط خليط من مدخنين وغير مدخنين. العمر يرفع التكلفة في كل الشرائط، لكن التدخين يرفعها أكثر بكثير.

### 3. Bar: متوسط التكاليف حسب التدخين وحسب المنطقة

```python
by_smoker = df.groupby("smoker")["charges"].mean()
by_region = df.groupby("region")["charges"].mean()
print(by_smoker.round(0))
print(by_region.round(0))

fig, axs = plt.subplots(1, 2, figsize=(12, 4))

axs[0].bar(by_smoker.index, by_smoker.values, color=["tab:blue", "tab:red"])
for i, v in enumerate(by_smoker.values):
    axs[0].text(i, v + 300, f"{v:,.0f}", ha="center")
axs[0].set_title("Average Charges by Smoking Status")
axs[0].set_xlabel("Smoker")
axs[0].set_ylabel("Average charges ($)")

axs[1].bar(by_region.index, by_region.values, color="#55A868")
for i, v in enumerate(by_region.values):
    axs[1].text(i, v + 300, f"{v:,.0f}", ha="center")
axs[1].set_title("Average Charges by Region")
axs[1].set_xlabel("Region")
axs[1].set_ylabel("Average charges ($)")

plt.tight_layout()
plt.show()
```

**شرح الكود:** `groupby("smoker")["charges"].mean()` تقسم الداتا حسب التدخين وتحسب متوسط التكاليف لكل مجموعة، ونفس الشي للمنطقة. `plt.subplots(1, 2)` تعطيني رسمتين جنب بعض. `by_smoker.index` هي أسماء المجموعات و`by_smoker.values` هي المتوسطات. `plt.text` تكتب الرقم فوق كل عمود.

**ماذا تقول هذه الرسمة؟** الفرق بين المدخن وغير المدخن ضخم: متوسط المدخن حوالي 32,000 مقابل حوالي 8,400 لغير المدخن، يعني قرابة 4 أضعاف. أما بين المناطق فالفروقات صغيرة نسبياً، وأعلى منطقة هي southeast وأقلها southwest. يعني التدخين عامل أقوى بكثير من المنطقة.

### 4. رسمة مضلِّلة عن قصد: المحور من 12,000 مقابل الصفر

```python
top = by_region.idxmax()
low = by_region.idxmin()
ratio = by_region.max() / by_region.min()
print(f"Highest: {top} ({by_region.max():,.0f}) | Lowest: {low} ({by_region.min():,.0f})")
print(f"Ratio highest/lowest = {ratio:.2f}")

fig, axs = plt.subplots(1, 2, figsize=(12, 4))

axs[0].bar(by_region.index, by_region.values, color="#C44E52")
axs[0].set_ylim(12000, by_region.max() + 300)
axs[0].set_title("Average Charges by Region (axis starts at 12,000)")
axs[0].set_xlabel("Region")
axs[0].set_ylabel("Average charges ($)")

axs[1].bar(by_region.index, by_region.values, color="#4C72B0")
axs[1].set_ylim(0, by_region.max() + 1500)
axs[1].set_title("Average Charges by Region (axis starts at 0)")
axs[1].set_xlabel("Region")
axs[1].set_ylabel("Average charges ($)")

plt.tight_layout()
plt.show()
```

**شرح الكود:** `idxmax()` و`idxmin()` ترجع اسم أعلى وأقل منطقة، وأحسب النسبة بقسمة الأعلى على الأقل. `set_ylim(12000, ...)` تخلي المحور يبدأ من 12,000 في الرسمة الأولى، و`set_ylim(0, ...)` تخليه يبدأ من الصفر في الثانية. نفس البيانات بالضبط، الفرق في بداية المحور فقط.

**ماذا تقول هذه الرسمة؟** في الرسمة الأولى (المحور من 12,000) يبان الفرق بين المناطق ضخم، وأعلى منطقة تبدو أضعاف أقلها، لأن الجزء المقطوع من الأعمدة يخفي حقيقة أنها كلها قريبة. أما في الرسمة الثانية (من الصفر) فالأعمدة شبه متقاربة، وهذا الحجم الحقيقي للفرق. النسبة الحقيقية بين أعلى منطقة وأقلها حوالي 1.2 (southeast إلى southwest)، يعني الأعلى أكبر بحوالي 20% بس، وهذا مو ضعف ولا قريب منه.

### محطة أرصاد المريخ

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(2026)
sol = np.arange(1, 61)
temp = np.round(-60 + 15 * np.sin(sol / 9) + rng.normal(0, 4, 60), 1)
storm = np.zeros(60, dtype=int)
storm[[11, 12, 13, 32, 33, 46, 47, 48, 49]] = 1   # 1 = عاصفة ترابية
temp[storm == 1] -= 12
wind = np.round(20 + storm * 35 + rng.normal(0, 5, 60), 1)

print("Storm sols:", sol[storm == 1])
```

**شرح الكود:** `sol` هي أيام المريخ من 1 إلى 60. `temp` حرارة كل يوم: موجة جيبية (`np.sin`) تمثل التغير الموسمي مع ضجيج عشوائي `rng.normal`. `storm` مصفوفة أصفار، وأحط 1 في أيام العواصف. بعدها أنزّل 12 درجة من حرارة أيام العواصف، وأرفع الرياح 35 فيها. `print` تعرض أرقام Sol اللي فيها عواصف عشان أتأكد.

### 1. الحرارة لكل Sol مع أيام العواصف

```python
plt.figure(figsize=(10, 4))
plt.plot(sol, temp, color="steelblue", label="Temperature")
plt.scatter(sol[storm == 1], temp[storm == 1], color="red", zorder=3, label="Dust storm")
plt.title("Mars Temperature by Sol")
plt.xlabel("Sol (Martian day)")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()
```

**شرح الكود:** `plt.plot` ترسم خط الحرارة. `sol[storm == 1]` تختار فقط أيام العواصف (فلترة بشرط)، و`temp[storm == 1]` حرارتها، و`plt.scatter` ترسمها نقاط حمراء فوق الخط. `zorder=3` يخلي النقاط فوق الخط وما تختفي تحته.

**ماذا تقول هذه الرسمة؟** الحرارة تتأرجح بشكل موجة حول -60 تقريباً، وأيام العواصف (النقاط الحمراء) تقع دائماً في الجزء الأبرد من الخط، لأن العاصفة تنزّل الحرارة حوالي 12 درجة. وأبرد نقاط الفترة كلها تقع في العواصف الأخيرة قرب Sol 47 إلى 50.

### 2. خط المتوسط وسهم لأبرد يوم

```python
mean_t = temp.mean()
i_cold = temp.argmin()

plt.figure(figsize=(10, 4))
plt.plot(sol, temp, color="steelblue", label="Temperature")
plt.scatter(sol[storm == 1], temp[storm == 1], color="red", zorder=3, label="Dust storm")
plt.axhline(mean_t, color="gray", linestyle="--", label=f"Mean = {mean_t:.1f}°C")
plt.annotate(f"Coldest: Sol {sol[i_cold]}, {temp[i_cold]}°C",
             xy=(sol[i_cold], temp[i_cold]),
             xytext=(sol[i_cold] - 22, temp[i_cold] + 10),
             arrowprops=dict(arrowstyle="->", color="black"))
plt.title("Mars Temperature with Mean and Coldest Day")
plt.xlabel("Sol (Martian day)")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()

print(f"Mean: {mean_t:.2f} | Coldest: Sol {sol[i_cold]} at {temp[i_cold]}°C")
```

**شرح الكود:** `temp.mean()` تحسب متوسط الحرارة، و`temp.argmin()` ترجع **فهرس** أبرد يوم (مو قيمته)، ومنه آخذ رقم الـ Sol وقيمة الحرارة. `plt.axhline` ترسم خط أفقي عند المتوسط. `plt.annotate` ترسم السهم: `xy` هو النقطة اللي يشير لها السهم، و`xytext` مكان النص، و`arrowprops` شكل السهم.

**ماذا تقول هذه الرسمة؟** متوسط الحرارة حوالي -60.7°C، والحرارة تدور حوله صعوداً ونزولاً. أبرد يوم هو Sol 48 بحرارة -87.5°C، وهو يوم عاصفة، وبعيد عن المتوسط بأكثر من 25 درجة.

### 3. الحرارة والرياح فوق بعض بنفس المحور الأفقي

```python
fig, axs = plt.subplots(2, 1, figsize=(10, 7), sharex=True)
fig.suptitle("Mars: Temperature and Wind", fontsize=14)

# الحرارة
axs[0].plot(sol, temp, color="steelblue", label="Temperature")
axs[0].scatter(sol[storm == 1], temp[storm == 1], color="red", zorder=3, label="Dust storm")
axs[0].set_title("Temperature")
axs[0].set_ylabel("Temperature (°C)")
axs[0].legend()

# الرياح: أحمر في أيام العواصف
colors = np.where(storm == 1, "red", "gray")
axs[1].bar(sol, wind, color=colors)
axs[1].set_title("Wind Speed (red = storm day)")
axs[1].set_xlabel("Sol (Martian day)")
axs[1].set_ylabel("Wind speed")

plt.tight_layout()
plt.show()
```

**شرح الكود:** `plt.subplots(2, 1, sharex=True)` تعطيني رسمتين فوق بعض، و`sharex=True` يخلي المحور الأفقي (Sol) مشترك بينهم، فأقدر أقارن نفس اليوم رأسياً. `np.where(storm == 1, "red", "gray")` تبني قائمة ألوان: أحمر لو اليوم عاصفة، رمادي لو لا. وبعدين `axs[1].bar(sol, wind, color=colors)` ترسم الأعمدة بهذي الألوان.

**ماذا تقول هذه الرسمة؟** كل يوم عاصفة تظهر فيه حرارة منخفضة في الرسمة العليا وعمود أحمر عالي في الرسمة السفلى، يعني العاصفة تجمع شيئين سيئين: برد أشد ورياح أقوى. متوسط الرياح في أيام العواصف حوالي 55 مقابل حوالي 19 في الأيام الهادية.

### 4. أفضل 5 أيام متتالية للإطلاق

**من الرسمة أولاً:** (اكتب هنا تخمينك من الرسمة قبل ما تشغّل الخلية اللي بعدها: أي 5 أيام تبدو أفضل؟)

```python
best_start = None
best_mean = -np.inf

for start in range(0, len(sol) - 4):
    window = slice(start, start + 5)
    if storm[window].sum() == 0:            # بلا عواصف
        m = temp[window].mean()
        if m > best_mean:                   # أدفأ نافذة لحد الآن
            best_mean = m
            best_start = start

win = slice(best_start, best_start + 5)
print(f"Best window: Sol {sol[best_start]} to Sol {sol[best_start + 4]}")
print(f"Average temperature: {best_mean:.1f}°C")
print(f"Wind in window: {wind[win]}")

# إثبات بالرسمة
plt.figure(figsize=(10, 4))
plt.plot(sol, temp, color="steelblue", label="Temperature")
plt.scatter(sol[storm == 1], temp[storm == 1], color="red", zorder=3, label="Dust storm")
plt.axvspan(sol[best_start] - 0.5, sol[best_start + 4] + 0.5, color="green", alpha=0.25, label="Best launch window")
plt.title("Best 5-Day Launch Window")
plt.xlabel("Sol (Martian day)")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()
```

**شرح الكود:** حلقة `for` تمر على كل يوم بداية ممكن (من 0 إلى 55)، وفي كل مرة `slice(start, start + 5)` تأخذ 5 أيام متتالية. الشرط `storm[window].sum() == 0` يعني ما في ولا عاصفة في هذي النافذة (لأن العاصفة = 1، فلو المجموع صفر ما في عواصف). لو النافذة آمنة أحسب متوسط حرارتها، ولو أعلى من أفضل متوسط سابق أحفظها. `plt.axvspan` تظلل النافذة الفائزة بالأخضر على الرسمة للإثبات.

**ماذا تقول هذه الرسمة؟** المنطقة الخضراء تقع في فترة بلا نقاط حمراء، وفي أعلى جزء من خط الحرارة بين العواصف، يعني أدفأ وأهدأ 5 أيام متتالية: من Sol 15 إلى Sol 19.
